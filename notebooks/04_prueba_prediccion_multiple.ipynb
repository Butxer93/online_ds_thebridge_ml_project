## 🎯 FUNCIÓN PRINCIPAL DE PREDICCIÓN

### Generando predicciones inteligentes

Esta es la función más importante: toma datos de entrada y produce predicciones sobre cómo responderá la empresa.

**¿Qué hace la predicción?**
1. Procesa los datos de entrada usando el pipeline
2. Alinea las características con lo que espera el modelo
3. Hace la predicción usando el modelo entrenado
4. Calcula niveles de confianza
5. Devuelve resultados estructurados

**¿Qué información retorna?**
- Tipo de respuesta predicha
- Nivel de confianza de la predicción
- Probabilidades para cada tipo de respuesta posible
- Información sobre la completitud de los datos

```python
if __name__ == "__main__":
    # Crear predictor optimizado
    predictor = RobustComplaintPredictor()

    # Ejemplos de prueba optimizados (ahora como lista)
    examples = [
        {'Product': 'Debt collection','Issue': 'Continued attempts to collect debt not owed and identity fraud concerns','State': 'NY','Company': 'Debt Collectors Inc','year_received': 2023,'Sub-product': 'Medical','Sub-issue': 'Debt is not mine','Consumer disputed?': 'Yes','processing_days': 5},
        {'Product': 'Mortgage'},
        {'Product': None, 'Issue': ''}
    ]
    
    print("="*80)
    print("🚀 SISTEMA OPTIMIZADO DE PREDICCIÓN DE QUEJAS")
    print("="*80)
    
    for i, example in enumerate(examples):
        print(f"\n🔍 PROCESANDO QUEJA {i+1}/{len(examples)}...")
        print(f"Input: {example}")
        
        result = predictor.predict(example)
        
        if result['prediction_successful']:
            # Mostrar top 3 probabilidades
            print(f"  Top probabilities:")
            sorted_probs = sorted(result['probabilities'].items(), 
                                key=lambda x: x[1], reverse=True)
            for response, prob in sorted_probs[:3]:
                print(f"    - {response}: {prob:.3f}")
        else:
            print(f"❌ Error: {result['error']}")
        
        print("-" * 50)

    examples.clear()
```
